```python

def calculate_iou(span1, span2):
    """Calcola l'IoU basato sui set di parole (tokens)."""
    tokens1 = set(span1.lower().split())
    tokens2 = set(span2.lower().split())
    
    intersection = tokens1.intersection(tokens2)
    union = tokens1.union(tokens2)
    
    if not union:
        return 0.0
    return len(intersection) / len(union)

def evaluate_with_iou(gold_spans, predicted_spans, threshold=0.10):
    """
    Valuta i TP basandosi sulla soglia IoU.
    Ogni Gold span deve trovare almeno un Predicted span con IoU > threshold.
    """
    if not gold_spans:
        return 0, 0, 0
    
   
            
 
    return tp, fn
```

```python
def compute_recall(data,threshold=0.50):

    tp,fn=0,0
    for item in data:
        pred_spans=item["pred"]
        gold_spans=item["gold"]
        for gold in gold_spans:
            
            best_iou = 0
            for pred in pred_spans:
                
                current_iou = calculate_iou(gold, pred)
                if current_iou >= best_iou:
                    best_iou = current_iou
            
            # Se il miglior match supera la soglia, è un successo
            if best_iou >= threshold:
                tp += 1
            else:
                fn += 1
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0
    return recall
```

```python
from glob import glob
import json
from collections import defaultdict
models=glob("FT/*.json")

for th in [0.10,0.50]:
    print("TH",th)
    res=defaultdict(float)
    for model_path in models:
        name=model_path.split("/")[-1].replace(".json","")
        data=json.load(open(model_path))
        recall=compute_recall(data,threshold=th)
        res[name]=recall
    print(res)
    
```
