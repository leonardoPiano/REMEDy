```python
from datasets import load_dataset
toxic_chat=load_dataset('lmsys/toxic-chat', 'toxicchat0124')["test"].to_pandas()

   
```

```python
human_toxic=toxic_chat[toxic_chat["human_annotation"]==True]
```

```python
import pandas as pd
import os
from glob import glob
from sklearn.metrics import accuracy_score, f1_score

import json
def accuracy_f1(real, preds):
    return {
        "ACC": round(accuracy_score(real, preds), 4),
        "F1": round(f1_score(real, preds), 4)
    }
def filter_preds(toxic_chat_pred):    
    filtered=[]
    human_prompts=human_toxic["user_input"].values.tolist()
    for item in toxic_chat_pred:
        
        if item["text"] in human_prompts:
            filtered.append(item)
    return filtered
def compute_scores(mode):
    
    base_path = f"../output/fold-0/parsed/{mode}/"
    models = [x for x in os.listdir(base_path) if not x.startswith(".")]
    
    for model in models:
        print("Model",model)
        toxic_chat_pred=json.load(open(f"{base_path}{model}/ToxicChat.json"))
        filtered=filter_preds(toxic_chat_pred)
        pred_df=pd.DataFrame(filtered)
        metrics = accuracy_f1(pred_df["real"], pred_df["pred"])
        print("Metrics")
        print(metrics)
```

```python
compute_scores("ZERO")
```
